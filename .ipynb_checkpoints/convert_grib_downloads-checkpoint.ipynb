```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import pandas as pd
from functools import reduce

# Path to your GRIB file
file_path = "Download 13.grib"

# Open the GRIB file using the cfgrib engine
try:
    ds = xr.open_dataset(file_path, engine="cfgrib")
    print(ds)
except Exception as e:
    print(f"Error opening file: {e}")
```

```python

# Step 1: Convert temperature from Kelvin to Fahrenheit
t2m_f = (ds['t2m'] - 273.15) * 9/5 + 32

# Step 2: Resample to daily minimum temperature
daily_min_f = t2m_f.resample(time='1D').min()

# Step 3: Apply land-sea mask (assume lsm=1 is land)
land_mask = ds['lsm'].isel(time=0) > 0.5  # static mask from first timestep
daily_min_land = daily_min_f.where(land_mask)

# Step 4: Group by year
grouped = daily_min_land.groupby('time.year')

# Helper: Drop non-essential coordinates before converting to DataFrame
def clean_coords(da):
    keep = ['latitude', 'longitude']
    if 'year' in da.coords:
        keep.append('year')
    return da.drop_vars([coord for coord in da.coords if coord not in keep], errors='ignore')

# Step 5: Compute metrics per year
def compute_days_below(grouped, threshold):
    da = grouped.apply(lambda x: (x < threshold).sum(dim='time'))
    da = da.assign_coords(year=da['year'])  # ensure year is a coordinate
    da = clean_coords(da)
    return da.to_dataframe(name=f'days_below_{threshold}F').reset_index()

# Compute thresholds in desired order
thresholds = [55, 32, 28, 26]
threshold_dfs = [compute_days_below(grouped, t) for t in thresholds]

# Compute annual minimum temperature
annual_min_temp = grouped.min(dim='time')
annual_min_temp = annual_min_temp.assign_coords(year=annual_min_temp['year'])
annual_min_temp = clean_coords(annual_min_temp)
annual_min_temp = annual_min_temp.to_dataframe(name='annual_min_temp_F').reset_index()

# Step 6: Merge all metrics
df = reduce(lambda left, right: left.merge(right, on=['year', 'latitude', 'longitude']),
            threshold_dfs + [annual_min_temp])

# Step 7: Rename and clean
df = df.rename(columns={'year': 'Year', 'latitude': 'Lat', 'longitude': 'Lon'})
df = df.dropna()

# Preview
print(df.head())

# Step 8: Export to CSV with year range in filename
year_min = df['Year'].min()
year_max = df['Year'].max()
output_path = f"annual_land_climate_summary_{year_min}_{year_max}.csv"

df.to_csv(output_path, index=False)
print(f"Data saved to {output_path}")
```
