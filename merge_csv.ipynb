```python
import os
import glob
import pandas as pd

# Use the current working directory (same folder as the notebook/script)
input_folder = os.getcwd()

# Find all CSV files in that folder
csv_files = glob.glob(os.path.join(input_folder, "*.csv"))

if not csv_files:
    raise FileNotFoundError("No CSV files found in the current folder.")

df_list = []
for file in csv_files:
    try:
        df = pd.read_csv(file)
        df_list.append(df)
        print(f"Loaded {os.path.basename(file)} with {len(df)} rows")
    except Exception as e:
        print(f"Skipping {file} due to error: {e}")

if df_list:
    combined_df = pd.concat(df_list, ignore_index=True)
    output_file = os.path.join(input_folder, "combined_output.csv")
    combined_df.to_csv(output_file, index=False)
    print(f"Combined CSV saved to: {output_file}")
else:
    print("No valid CSV files to merge.")
```
